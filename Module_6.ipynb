### Подвыборка только людей от 20 до 30

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('titanic.csv')
df.head()

df_age = df[(df['Age'] >= 20) & (df['Age'] <= 30)]
df_age.head()

print('Количество пассажиров:', len(df_age))
```

### Подсчет базовой статистики (среднее, медиану, минимум и максимум) цены за билет у этой подвыборки людей

```python
print('Средняя цена билета:', df_age['Fare'].mean())

print('Медиана:', df_age['Fare'].median())

print('Минимальная цена:', df_age['Fare'].min())

print('Максимальная цена:', df_age['Fare'].max())

df_age['Fare'].agg(['mean', 'median', 'min', 'max'])
```

### График распределения цены за билет по полу в этой подвыборке

```python
plt.figure(figsize=(10, 6))

sns.violinplot(
    data=df_age,
    x='Sex',
    y='Fare',
    cut=0,
    inner=None,
    color='skyblue'
)

sns.stripplot(
    data=df_age,
    x='Sex',
    y='Fare',
    color='black',
    alpha=0.35,
    jitter=True
)

plt.title('Стоимость билетов мужчин и женщин (20–30 лет)')
plt.xlabel('Пол')
plt.ylabel('Цена билета')

plt.ylim(0, df_age['Fare'].max() * 1.05)
plt.tight_layout()
plt.show()
```
