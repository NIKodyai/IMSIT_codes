### Список students из десяти имен

```python
students = ['Алексей', 'Мария', 'Дмитрий', 'Анна', 'Сергей', 'Елена', 'Максим', 'Ольга', 'Иван', 'Татьяна']
students = list(sorted(set(students), key=students.index))
print(students)
```

### Поиск первого элемента списка

### Проверка, есть ли латинский символ ‘a’ в имени предпоследнего студента

```python
students = ['Алексей', 'Мария', 'Дмитрий', 'Анна', 'Сергей', 'Елена', 'Максим', 'Ольга', 'Иван', 'Татьяна']
penultimate_student = students[-2]
has_latin_a = "а" in penultimate_student

print("Предпоследний студент:", penultimate_student)
print("Есть ли латинский символ 'a':", has_latin_a)
```
