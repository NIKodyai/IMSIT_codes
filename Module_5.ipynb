### Загрузка файла titanic.csv

```python
import pandas as pd
import shutil
from pathlib import Path

file_path = Path('titanic.csv')

# Создаём резервную копию исходного файла
backup_path = Path('titanic_backup.csv')
if not backup_path.exists():
    shutil.copy2(file_path, backup_path)

# Загружаем данные
df = pd.read_csv(file_path)
print('Размер исходной таблицы:', df.shape)
df.head()
```

### Запись в этот файл любые 3 строки с данными

```python
three_rows = df.head(3)
three_rows

with open(file_path, mode='w', encoding='utf-8', newline='') as file:
    three_rows.to_csv(file, index=False)

print('Три строки записаны. Файл закрыт.')
```
