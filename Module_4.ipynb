### Создание списка students

```python
students = ['Алексей', 'Мария', 'Дмитрий', 'Анна', 'Сергей', 'Елена', 'Максим', 'Ольга', 'Иван', 'Татьяна',]
print("Список students:", students)
```

### Функция students, имя начинается с гласной - имя полностью заменяется на invalid

```python
VOWELS = set("аеёиоуыэюяaeiouАЕЁИОУЫЭЮЯAEIOU")

def filter_students(student_list):

  return [
      "invalid" if name and name[0] in VOWELS else name for name in student_list
  ]

filter_students = filter_students(students)
print("Обновлённый список students:", filter_students)
```

### Функция подсчета 'invalid' в любом списке

```python
count_invalid = filter_students.count("invalid")
print("Количество 'invalid' в students:", count_invalid)
```
