# Задание 2

```python
print("--- Задание 2: Сложение двух чисел ---")
a = float(input("Введите первое число: "))
b = float(input("Введите второе число: "))
result = a + b
print(f"Результат: {a} + {b} = {result}")
```

# Задание 3

```python
print("--- Задание 3: Деление двух чисел ---")
a = float(input("Введите делимое: "))
b = float(input("Введите делитель: "))

if b == 0:
  print("Ошибка: деление на ноль невозможно!")
else:
  result = a / b
  print(f"Результат: {a} / {b} = {result}")
```

# Задание 4

```python
print("--- Задание 4: Возведение в степень ---")
base = float(input("Введите число (основание): "))
exponent = float(input("Введите степень: "))
result = base**exponent
print(f"Результат: {base} в степени {exponent} = {result}")
```
