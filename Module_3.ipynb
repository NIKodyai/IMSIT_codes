### 2 множества студентов

```python
group1 = {'Алексей', 'Мария', 'Дмитрий', 'Анна', 'Сергей'}
group2 = {'Мария', 'Елена', 'Дмитрий', 'Максим', 'Ольга'}
print("Множество 1:", group1)
print("Множество 2:", group2)
```

### Пересечение множеств и сохранение в списке

```python
intersection = group1.intersection(group2) # intersection = list(group_1 & group_2)

print("Пересечение множеств (список):", intersection)
```

### Объединение множеств и сохранение в списке

```python
union_list = list(group1 | group2)

print("Объединение множеств (список):", union_list)
```
